```python
import numpy as np
```

```python
random_integers = np.random.randint(1, 101, size=10)

print("Random Integers:")
print(random_integers)
```

```python
#Generate Random Floating-Point Numbers
random_floats = np.random.rand(10)

print("Random Floating-Point Numbers:")
print(random_floats)
```

```python
#Generate Random Numbers Within a Range
random_numbers = np.random.uniform(10, 50, size=10)

print("Random numbers between 10 and 50:")
print(random_numbers)
```

```python
#Create a Small Simulated Dataset
student_marks = np.random.randint(35, 101, size=10)

print("Student Marks:")
print(student_marks)
```

```python
#Take Random Samples From an Array
students = np.array([
    "amir", "Priti", "Rahul", "Neha", "Geeha",
    "Rohan", "Pooja", "Akash", "Kiran", "Meena"
])

sample_students = np.random.choice(students, size=5, replace=False)

print("Random Student Sample:")
print(sample_students)
```

```python
#Demonstrate Random Seed
np.random.seed(42)

numbers1 = np.random.randint(1, 101, size=5)

print("First result:")
print(numbers1)
```

```python
#Demonstrate Reproducibility
np.random.seed(42)

result1 = np.random.rand(5)

np.random.seed(42)

result2 = np.random.rand(5)

print("Result 1:")
print(result1)

print("\nResult 2:")
print(result2)

print("\nAre both results same?", np.array_equal(result1, result2))
```

```python
#Generate a Random Student Dataset
np.random.seed(42)

student_ids = np.arange(1, 11)
marks = np.random.randint(35, 101, size=10)
attendance = np.random.randint(60, 101, size=10)

print("Student IDs:")
print(student_ids)

print("\nMarks:")
print(marks)

print("\nAttendance:")
print(attendance)
```

```python
#Display Dataset
student_data = np.column_stack((student_ids, marks, attendance))

print("Student Dataset:")
print(student_data)
```

```python
#Random Choice From Dataset
np.random.seed(42)

selected_marks = np.random.choice(marks, size=5, replace=False)

print("Randomly Selected Marks:")
print(selected_marks)
```

```python
#Final Example
np.random.seed(42)

data = np.random.randint(1, 101, size=20)

print("Generated Random Dataset:")
print(data)

sample = np.random.choice(data, size=5, replace=False)

print("\nRandom Sample:")
print(sample)
```
